```python
from pyspark.sql import functions as F
dfMonthlySalesClean = dfMonthlySales.selectExpr('UPPER(TRIM(sale_id)) as sale_id', 'UPPER(TRIM(product)) as product', 'UPPER(TRIM(category)) as category', 'TRIM(quantity) as quantity', 'TRIM(price) as price','TRIM(sale_date) as sale_date', 'Upper(TRIM(region)) as region','current_date() as ingest_ts')
#dfMonthlySalesClean.display()
dfMonthlySalesClean.write.mode("overwrite").option("mergeSchema", "true").saveAsTable("intellibi_catalog.intellibi_bronze.dfMonthlySales")
```
